```python
# Test with pymysql (same library backend uses)
import pymysql
import time

db_config = {
    'host': '192.168.1.149',
    'port': 3307,  # MUST be int
    'user': 'CBSUSER',
    'password': 'Falcon@2022',
    'database': 'neo'
}


# db_config = {
#     'host': '127.0.0.1',
#     'port': 3307,  # MUST be int
#     'user': 'falcon_ai',
#     'password': 'falcon@ai@123',
#     'database': 'neo'
# }
print("Testing pymysql connection (same as backend)...")
print(f"Config: {db_config}")

try:
    start = time.time()
    conn = pymysql.connect(
        host=db_config['host'],
        port=db_config['port'],
        user=db_config['user'],
        password=db_config['password'],
        database=db_config['database'],
        charset='utf8mb4',
        connect_timeout=15,
        cursorclass=pymysql.cursors.DictCursor
    )
    elapsed = time.time() - start
    print(f"✅ Connected successfully in {elapsed:.2f}s")
    
    with conn.cursor() as cur:
        cur.execute("SELECT DATABASE(), VERSION()")
        result = cur.fetchone()
        print(f"Database: {result['DATABASE()']}")
        print(f"Version: {result['VERSION()']}")
    
    conn.close()
    print("✅ Connection closed")
    
except Exception as e:
    print(f"❌ Connection failed: {e}")
    print(f"Error type: {type(e).__name__}")
```

```python
# receiver.py
import socket

s = socket.socket()
s.bind(("0.0.0.0", 5001))
s.listen(1)

conn, _ = s.accept()

with open("received.bin", "wb") as f:
    while True:
        data = conn.recv(4096)
        if not data:
            break
        f.write(data)

conn.close()
```
